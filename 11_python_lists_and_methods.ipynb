# Lists and List Methods

A list in Python is an **ordered, mutable** collection of elements. It can contain elements of different types.

## Creating a List

```python
# Empty list
my_list = []

# List with elements
numbers = [1, 2, 3, 4, 5]

# Mixed data types
mixed_list = [1, "Hello", 3.14, True]
print(mixed_list)
```

## Common List Methods

| Method | Description |
|---|---|
| `append(x)` | Adds an element `x` to the **end** of the list |
| `extend(iterable)` | Appends all elements from an iterable |
| `insert(index, x)` | Inserts `x` at the specified `index` |
| `remove(x)` | Removes the **first occurrence** of `x` |
| `pop([index])` | Removes and returns the element at `index` (last element if not provided) |
| `index(x)` | Returns the index of the first occurrence of `x` |
| `count(x)` | Returns how many times `x` appears |
| `sort()` | Sorts the list in ascending order (in place) |
| `reverse()` | Reverses the order of the list (in place) |
| `copy()` | Returns a shallow copy of the list |
| `clear()` | Removes all elements |

```python
fruits = ["apple", "banana", "cherry"]
fruits.append("orange")
print(fruits)  # ['apple', 'banana', 'cherry', 'orange']

fruits.sort()
print(fruits)  # ['apple', 'banana', 'cherry', 'orange']
```

```python
my_list = [1, 2, 3]

my_list.extend([4, 5])        # Add multiple elements
print(my_list)

my_list.insert(1, "Python")   # Insert at index 1
print(my_list)

my_list.remove("Python")      # Remove first occurrence
print(my_list)

last = my_list.pop()          # Remove and return last element
print("Popped:", last, "→", my_list)

print("Index of 3:", my_list.index(3))
print("Count of 2:", my_list.count(2))

copy = my_list.copy()         # Shallow copy
my_list.clear()               # Remove all elements
print("Cleared:", my_list, "| Copy survived:", copy)
```

## Indexing and Slicing

```python
nums = [10, 20, 30, 40, 50]

print(nums[0])      # First element → 10
print(nums[-1])     # Last element → 50
print(nums[1:4])    # Slice → [20, 30, 40]
print(nums[:3])     # First three → [10, 20, 30]
print(nums[::2])    # Every second element → [10, 30, 50]
print(nums[::-1])   # Reversed → [50, 40, 30, 20, 10]
```

## Useful Built-in Functions

`len()`, `min()`, `max()`, `sum()`, `sorted()` — and note the difference between `sort()` and `sorted()`:

```python
scores = [85, 40, 75, 60]

print("Length:", len(scores))
print("Min:", min(scores), "| Max:", max(scores), "| Sum:", sum(scores))

# sorted() returns a NEW list, original is unchanged
ascending = sorted(scores)
descending = sorted(scores, reverse=True)
print("Sorted (new):", ascending)
print("Original still:", scores)

# sort() modifies the list IN PLACE and returns None
scores.sort(reverse=True)
print("After .sort():", scores)
```

## Sorting with a Custom `key`

```python
words = ["cherry", "fig", "banana"]
words.sort(key=len)  # Sort by length
print(words)  # ['fig', 'banana', 'cherry']
```

## List Operations: `+`, `*`, and `in`

```python
a = [1, 2]
b = [3, 4]

print(a + b)   # Concatenation → [1, 2, 3, 4]
print(a * 3)   # Repetition → [1, 2, 1, 2, 1, 2]
print(2 in a)  # Membership → True
```

## ⚠️ Common Pitfall: Copy vs. Reference

Assigning one list to another creates a **reference**, not a copy — both names point to the same list.

```python
original = [1, 2, 3]
alias = original          # Same list, two names
real_copy = original.copy()  # Independent copy

alias.append(99)
print("Alias changed original:", original)  # [1, 2, 3, 99]
print("Copy unaffected:", real_copy)         # [1, 2, 3]
```

## Nested Lists

```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(matrix[1][2])  # Row 1, Column 2 → 6
```

## Lists in Data Science

### Example: Basic Statistics with a List of Scores

```python
scores = [85, 40, 75, 60, 92, 55]

average = sum(scores) / len(scores)
top_scores = sorted(scores, reverse=True)[:3]  # Top 3
passed = [s for s in scores if s >= 50]        # Filter with list comprehension

print(f"Average: {average:.1f}")
print("Top 3:", top_scores)
print(f"Passed {len(passed)}/{len(scores)}: {passed}")
```

## Summary

- Lists are **ordered and mutable**, and can hold mixed types.
- Key methods: `append`, `extend`, `insert`, `remove`, `pop`, `sort`, `copy`, `clear`.
- `sort()` modifies in place; `sorted()` returns a new list.
- Slicing with `[start:stop:step]` is essential for working with list data.
- Use `.copy()` (or `list()`) when you need an independent copy.
