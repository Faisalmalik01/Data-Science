# 04 — Taking Input from the User

## 1. Basic usage of `input()`

The built-in `input()` function displays a prompt and waits for the user to type something. When the user presses Enter, Python returns what they typed.

**Important:** `input()` always returns a string (`str`), even if the user types digits.

```python
# Ask the user to type their name.
name = input("Enter your name: ")

# Display the text entered by the user.
print("Hello,", name)
```

## 2. Why does `input()` return a string?

If the user types `25`, Python receives the characters `"25"` as text—not the integer `25`.

That means adding two inputs directly joins the strings together instead of doing numeric addition.

```python
# Ask the user to enter two values.
first = input("Enter the first number: ")
second = input("Enter the second number: ")

# Both values are strings, so + concatenates them.
print(first + second)
```

**Try it:** enter `2` for the first value and `3` for the second. The result will be `23`, not `5`, because both inputs are strings.

## 3. Convert input into numbers

Use `int()` when you expect a whole number and `float()` when you expect a decimal number.

- `int("25")` becomes the integer `25`.
- `float("3.5")` becomes the floating-point number `3.5`.

Conversion fails if the text is not a valid number for that type. For example, `int("hello")` raises a `ValueError`.

```python
# Ask for two whole numbers and convert each input to int.
first = int(input("Enter the first whole number: "))
second = int(input("Enter the second whole number: "))

# Now + performs arithmetic addition.
print("Sum:", first + second)
```

```python
# Ask for a decimal number and convert the input to float.
price = float(input("Enter a price: "))

# Multiply the price by 2.
print("Double the price:", price * 2)
```

## 4. Use input in calculations

Once input is converted to a number, you can use it in arithmetic expressions.

```python
# Get a person's age as a whole number.
age = int(input("Enter your age: "))

# Calculate the age they will be next year.
next_year_age = age + 1

# Display the result.
print("Next year, you will be", next_year_age)
```

## 5. Taking multiple inputs

You can ask for several values one after another. Convert each value to the appropriate type if you need to calculate with it.

```python
# Ask for a student's name and marks.
student_name = input("Enter the student's name: ")
marks = float(input("Enter the marks: "))

# Display the values using an f-string.
print(f"{student_name} scored {marks} marks.")
```

## 6. Taking two values on one line

You can use `.split()` to separate input at whitespace. For two whole numbers, `map(int, ...)` converts both pieces of text to integers.

```python
# The user should type two whole numbers separated by a space, e.g. 10 20.
a, b = map(int, input("Enter two whole numbers separated by a space: ").split())

# Display their sum.
print("Sum:", a + b)
```

If you type more or fewer than two values in the previous example, Python will raise an error because the input must provide exactly two values for `a` and `b`.

## 7. Input and conditions

User input can be used in an `if` statement. In this example, we convert the input to an integer before comparing it with `18`.

```python
# Ask for an age and convert it to an integer.
age = int(input("Enter your age: "))

# Check whether the age is at least 18.

if age >= 18:
    print("You are 18 or older.")
else:
    print("You are younger than 18.")
```

## 8. Common beginner mistakes

- **Forgetting conversion:** `input()` returns a string, so convert it before numeric calculations.
- **Using `int()` for decimal text:** `int("3.5")` raises a `ValueError`; use `float("3.5")` if a decimal is expected.
- **Typing non-numeric text when a number is expected:** `int("abc")` raises a `ValueError`.
- **Forgetting that input waits:** the program pauses at `input()` until the user submits a response.
- **Confusing the prompt with the stored value:** the prompt is the message inside `input("...")`; the user's response is stored in the variable on the left of `=`.
