# Python Operator Precedence (PEMDAS)

## 🔢 Order of Operations

Python follows **PEMDAS**:

1. **Parentheses `()`** → Highest precedence, operations inside parentheses are evaluated first.
2. **Exponents `**`** → Power calculations
3. **Multiplication `*`, Division `/`, Floor Division `//`, Modulus `%`** → Left to right
4. **Addition `+`, Subtraction `-`** → Evaluated from left to right

## 1. Parentheses

Operations inside parentheses are evaluated first.

```python
result = (10 + 2) * 3
print("(10 + 2) * 3 =", result)
```

## 2. Exponents

Exponentiation is evaluated right-to-left.

```python
result = 2 ** 3 ** 2  # Equivalent to 2 ** (3 ** 2)
print("2 ** 3 ** 2 =", result)
```

## 3. Multiplication, Division, Floor Division, Modulus

These are evaluated from left to right.

```python
print("10 / 3 =", 10 / 3)    # Division
print("10 // 3 =", 10 // 3)  # Floor division
print("10 % 3 =", 10 % 3)    # Modulus
print("2 * 3 =", 2 * 3)      # Multiplication
```

## 4. Addition and Subtraction

Finally, addition and subtraction are evaluated left to right.

```python
result = 10 + 2 * 3  # Multiplication happens first
print("10 + 2 * 3 =", result)

result = 20 - 5 + 2  # Left to right
print("20 - 5 + 2 =", result)
```

## ✅ Summary

Python strictly follows **PEMDAS**:

1. **P**arentheses
2. **E**xponents
3. **M**ultiplication / **D**ivision / Floor Division / Modulus
4. **A**ddition / **S**ubtraction

Understanding precedence helps avoid mistakes and ensures correct results.
