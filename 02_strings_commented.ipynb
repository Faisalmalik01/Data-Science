# 02 — Python Strings

A **string** (`str`) is a sequence of characters used to represent text. Examples include names, messages, filenames, and sentences.

Strings can use single quotes (`'...'`) or double quotes (`"..."`). Choose the style that makes the text easiest to write.

```python
# Store a name as a string.
name = "Faisal"

# In Jupyter, a variable written as the final line displays its value.
name
```

## 1. Changing letter case

`.upper()` returns a new string with letters converted to uppercase. `.lower()` returns a lowercase version.

These methods do not change the original string; strings are **immutable** in Python.

```python
# Convert the name to uppercase.
print(name.upper())

# Convert the name to lowercase.
print(name.lower())

# The original string remains unchanged.
print(name)
```

## 2. Checking how a string ends

`.endswith("text")` checks whether a string ends with the given text. It returns a Boolean: `True` or `False`.

The check is case-sensitive, so `"Faisal".endswith("al")` is `True`, but `"Faisal".endswith("AL")` is `False`.

```python
# Does name end with the characters "al"?
print(name.endswith("al"))

# Does name end with the characters "sal"?
print(name.endswith("sal"))

# The check is case-sensitive.
print(name.endswith("AL"))
```

## 3. Counting characters or matches

- `len(string)` returns the number of characters in the string.
- `.count("a")` counts non-overlapping occurrences of the substring `"a"`.

Spaces and punctuation count toward `len()`.

```python
# Count how many lowercase "a" characters occur in name.
print(name.count("a"))

# Count all characters in name.
print(len(name))

# Spaces also count as characters.
print(len("hi there"))
```

## 4. Removing surrounding whitespace with `.strip()`

`.strip()` removes whitespace from the **beginning and end** of a string. It does not remove spaces between words.

```python
# This string has spaces before and after the word.
s = " hello "

# Remove surrounding whitespace and display the result.
print(s.strip())

# The original string still contains its spaces.
print(repr(s))
```

## 5. Splitting a string with `.split()`

`.split(separator)` breaks a string into a list of smaller strings wherever the separator occurs.

For example, splitting a comma-separated string on `","` creates a list. Notice that spaces after commas remain in the individual items; you can clean them with `.strip()`.

```python
# A string containing items separated by commas.
s1 = "apple, banana, grape"

# Split wherever a comma occurs.
fruits = s1.split(",")

print(fruits)

# Remove surrounding spaces from each item.
clean_fruits = [fruit.strip() for fruit in fruits]
print(clean_fruits)
```

## 6. Indexing: accessing one character

Strings are indexed from **0**:

- index `0` is the first character
- index `1` is the second character
- index `-1` is the last character

Trying to access an index outside the string's range raises an `IndexError`.

```python
# Store the word Python in a string.
text = "Python"

# Access individual characters by index.
print(text[0])   # P: first character
print(text[1])   # y: second character
print(text[-1])  # n: last character
```

## 7. Slicing: accessing a portion of a string

The general form is `text[start:stop]`.

- `start` is included.
- `stop` is excluded.
- If `start` is omitted, slicing starts at the beginning.
- If `stop` is omitted, slicing continues to the end.

```python
# Get characters from index 0 up to, but not including, index 5.
print(name[0:5])

# From the beginning up to index 3 (index 3 is excluded).
print(text[:3])

# From index 2 through the end.
print(text[2:])

# Copy the whole string using slicing.
print(text[:])
```

## 8. More useful string operations

These are common operations you will use when cleaning user input or working with text data.

```python
message = "  Learn Python  "

# Remove spaces at both ends.
cleaned = message.strip()
print(cleaned)

# Replace one piece of text with another.
print(cleaned.replace("Python", "Data Science"))

# Check whether a string starts with a prefix.
print(cleaned.startswith("Learn"))

# Find the position of a substring; -1 means it was not found.
print(cleaned.find("Python"))
```

## 9. Combining strings

Use `+` to concatenate strings. Use an f-string when you want to insert variable values into text.

```python
first_name = "Faisal"
subject = "Python"

# Concatenation joins strings together.
print(first_name + " is learning " + subject)

# An f-string is often easier to read.
print(f"{first_name} is learning {subject}.")
```

## 10. Common beginner mistakes

- **Confusing index and length:** `"Python"` has length `6`, but its indexes are `0` through `5`.
- **Forgetting the stop index is excluded:** `"Python"[0:2]` gives `"Py"`.
- **Expecting `.strip()` to remove inner spaces:** `"a  b".strip()` remains `"a  b"`.
- **Expecting methods to alter the original string:** assign the returned value if you want to keep the transformed version, e.g. `name = name.lower()`.
- **Case sensitivity:** `"A"` and `"a"` are different characters.

## Quick recap

- A string stores text.
- `.upper()` and `.lower()` change letter case in the returned string.
- `.endswith()` and `.startswith()` return Boolean values.
- `.count()` counts matches; `len()` counts characters.
- `.strip()` removes surrounding whitespace.
- `.split()` separates text into a list.
- Indexing selects one character; slicing selects a range.
- Strings are immutable, so methods return new strings rather than editing the original in place.
