# 01 — Python Basics: Introduction

## 1. Variables

A **variable** is a name that refers to a value. The `=` symbol assigns a value to a variable.

```python
age = 20
```
Here, `age` is the variable name and `20` is the value. Python does not require you to declare a variable's type in advance.

````python
# Store the integer 5 in a variable named a.
a = 5

# Store the integer 32 in a variable named b.
b = 32

# Display the current value of a.
a
````

## 2. Using variables in an expression

Python evaluates `a + b` and, in a notebook, displays the result when it is the final expression in a cell.

````python
# Add the values stored in a and b.
a + b
````

## 3. Strings and variable names

Text is represented by a **string**. Strings can be written using single quotes or double quotes.

Variable names are case-sensitive: `name`, `Name`, and `NAME` are different names.

````python
# Store a text value (a string) in the variable name.
name = "Faisal"

# Display the value of name.
name
````

## 4. Displaying output with `print()`

`print()` displays a value or message. It is useful in scripts and works in notebooks too.

A notebook may automatically display the result of the last expression in a cell; `print()` explicitly asks Python to display it.

````python
# Display the value stored in name.
print(name)

# Display a literal message.
print("Hello, world!")
````

## 5. Checking data types with `type()`

A **data type** tells Python what kind of value something is.

- `int`: whole numbers, such as `5`
- `float`: decimal numbers, such as `3.14`
- `str`: text, such as `"hello"`
- `bool`: Boolean values, `True` or `False`

`type(value)` reports the type of a value.

````python
# Check the type of a, which currently stores an integer.
print(type(a))

# Check the type of name, which stores text.
print(type(name))

# Try a few other types.
print(type(3.14))      # float
print(type(True))      # bool
````

## 6. A number written as text is still text

The values below contain digits, but quotation marks make them strings (`str`), not integers.

When two strings are joined with `+`, Python performs **concatenation** rather than arithmetic addition.

````python
# These are strings because the values are inside quotation marks.
a = "4566"
b = "5677"

# Join the two strings together; this is not numeric addition.
print(a + b)

# Check their types to confirm they are strings.
print(type(a))
print(type(b))
````

## 7. Type conversion (type casting)

`int()` converts a suitable string of digits into an integer. After conversion, `+` performs arithmetic addition.

Not every string can be converted to an integer: for example, `int("hello")` raises a `ValueError`.

````python
# Convert each numeric string to an integer, then add the numbers.
print(int(a) + int(b))

# You can also convert an integer to a string.
number = 25
number_as_text = str(number)
print(number_as_text)
print(type(number_as_text))
````

## 8. Important distinction: `=` versus `==`

- `=` assigns a value to a variable.
- `==` compares two values and produces `True` or `False`.

````python
# Assignment: store 10 in x.
x = 10

# Comparison: is x equal to 10?
print(x == 10)  # True

# Comparison: is x equal to 7?
print(x == 7)   # False
````

## 9. Common beginner mistakes

- **Forgetting quotes:** `name = Faisal` makes Python look for a variable called `Faisal`; use `name = "Faisal"` for text.
- **Expecting string addition to add numbers:** `"2" + "3"` gives `"23"`, while `int("2") + int("3")` gives `5`.
- **Confusing `print()` with `type()`:** `print()` displays a value; `type()` tells you its type.
- **Invalid conversion:** `int("3.5")` fails; use `float("3.5")` if a decimal number is intended.
