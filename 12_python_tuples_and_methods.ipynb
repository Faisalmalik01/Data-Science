# Tuples and Tuple Methods

A tuple in Python is an **ordered, immutable** collection of elements. It is similar to a list, but once created, its elements **cannot be modified**.

## Creating a Tuple

```python
# Empty tuple
empty_tuple = ()

# Tuple with elements
numbers = (1, 2, 3, 4, 5)

# Mixed data types
mixed_tuple = (1, "Hello", 3.14, True)

# Single element tuple (comma is necessary!)
single_element = (42,)
not_a_tuple = (42)   # Just an int in parentheses

print(type(single_element), type(not_a_tuple))
```

## Tuple Characteristics

- **Immutable**: Once created, elements cannot be changed.
- **Faster than lists**: Accessing elements in a tuple is faster than in a list.
- **Can be used as dictionary keys**: Since tuples are immutable, they can be keys in dictionaries (lists cannot).

```python
# Immutability in action
my_tuple = (1, 2, 3)
# my_tuple[0] = 99   # ❌ TypeError: 'tuple' object does not support item assignment

# But you can always create a NEW tuple
new_tuple = my_tuple + (4, 5)
print(new_tuple)
```

## Accessing Tuple Elements

```python
my_tuple = (10, 20, 30, 40)

# Indexing
print(my_tuple[1])     # 20
print(my_tuple[-1])    # 40 (last element)

# Slicing
print(my_tuple[1:3])   # (20, 30)
print(my_tuple[::-1])  # (40, 30, 20, 10) — reversed
```

## Tuple Methods

Tuples have only **two** methods (because they're immutable):

| Method | Description | Example |
|---|---|---|
| `count(x)` | Returns the number of times `x` appears | `my_tuple.count(2)` |
| `index(x)` | Returns the index of the first occurrence of `x` | `my_tuple.index(3)` |

```python
nums = (1, 2, 2, 3, 2)

print("Count of 2:", nums.count(2))  # 3
print("Index of 3:", nums.index(3))  # 3
```

## Useful Built-in Functions and Operations

```python
scores = (85, 40, 75)

print("Length:", len(scores))
print("Min:", min(scores), "| Max:", max(scores), "| Sum:", sum(scores))
print("40 in scores:", 40 in scores)        # Membership
print("Concatenate:", scores + (60,))       # + creates a new tuple
print("Repeat:", (1, 2) * 3)               # (1, 2, 1, 2, 1, 2)

# Converting between list and tuple
print("From list:", tuple([1, 2, 3]))
print("To list:", list((1, 2, 3)))
```

## Tuple Packing and Unpacking

```python
# Packing
person = ("Alice", 25, "Engineer")

# Unpacking
name, age, profession = person
print(name)       # Alice
print(age)        # 25
print(profession) # Engineer
```

```python
# Unpacking with *rest
first, *middle, last = (1, 2, 3, 4, 5)
print(first, middle, last)  # 1 [2, 3, 4] 5

# Classic swap idiom — no temporary variable needed
a, b = 5, 10
a, b = b, a
print(f"a={a}, b={b}")  # a=10, b=5
```

## Tuples as Dictionary Keys

Because tuples are immutable, they can be used as dictionary keys — useful for coordinate grids or lookup tables (a list would raise a `TypeError`).

```python
city_population = {
    ("USA", "New York"): 8_300_000,
    ("Japan", "Tokyo"): 13_960_000,
    ("France", "Paris"): 2_100_000,
}

print(city_population[("Japan", "Tokyo")])
```

## List vs. Tuple

| Feature | List `[ ]` | Tuple `( )` |
|---|---|---|
| Mutable | ✅ Yes | ❌ No |
| Speed | Slower | Faster |
| Methods | Many | Only `count`, `index` |
| Dictionary keys | ❌ No | ✅ Yes |
| Use case | Data that changes | Fixed records, coordinates |

## Tuples in Data Science

### Example: Fixed Records (Coordinates) and Unpacking

```python
# Storing records that should never change
data_points = [("Alice", 85), ("Bob", 40), ("Charlie", 75)]

for name, score in data_points:   # Unpacking in a loop
    print(f"{name}: {score}")
```

## When to Use Tuples?

- When you want an **unchangeable** collection of elements.
- When you need a **faster** alternative to lists.
- When storing **heterogeneous data** (e.g., database records, coordinates).
