# Functions in Python

A **function** is a reusable block of code that performs a specific task. Functions help organize code, avoid repetition, and make programs easier to test and maintain — essential skills in data science for building reusable data processing pipelines.

## Defining a Function

Functions are defined using the `def` keyword, followed by the function name and parentheses.

```python
def greet():
    print("Hello, welcome to Python!")

greet()  # Calling the function
```

## Functions with Parameters

Parameters allow you to pass data into a function.

```python
def greet(name):
    print(f"Hello, {name}!")

greet("Alice")
greet("Bob")
```

## The `return` Statement

`return` sends a value back to the caller. A function without `return` returns `None` by default.

```python
def add(a, b):
    return a + b

result = add(3, 4)
print("Sum:", result)
```

## Default Parameters

You can give parameters default values, making them optional when calling the function.

```python
def greet(name="Guest"):
    print(f"Hello, {name}!")

greet()          # Uses the default value
greet("Charlie") # Overrides the default
```

## Keyword and Multiple Arguments

You can pass arguments by name, and accept any number of arguments using `*args`.

```python
def describe_person(name, age):
    print(f"{name} is {age} years old.")

describe_person(age=25, name="Diana")  # Keyword arguments (order doesn't matter)

def total(*args):
    return sum(args)

print("Total:", total(1, 2, 3, 4))
```

## Lambda Functions

Lambda functions are small anonymous functions defined in a single line — very common in data science with `pandas` and `map()`.

```python
square = lambda x: x ** 2
print("Square of 5:", square(5))

nums = [1, 2, 3, 4]
squared = list(map(lambda x: x ** 2, nums))
print("Squared list:", squared)
```

## Functions in Data Science

Functions are used to encapsulate data cleaning, feature engineering, and reusable transformations.

### Example: Data Cleaning Function

```python
import pandas as pd

def categorize_score(score):
    if score >= 80:
        return "A"
    elif score >= 60:
        return "B"
    elif score >= 40:
        return "C"
    else:
        return "F"

data = {'Name': ['Alice', 'Bob', 'Charlie', 'Diana'], 'Score': [85, 40, 75, 55]}
df = pd.DataFrame(data)
df['Grade'] = df['Score'].apply(categorize_score)
print(df)
```

## Summary

- **`def`** defines a reusable block of code.
- **Parameters** pass data in; **`return`** sends results back.
- **Default parameters** make arguments optional.
- **Lambda functions** provide quick one-line functions.
- Functions make data science workflows reusable, testable, and clean.
