# Python Dictionaries and Dictionary Methods
A beginner-friendly notebook covering dictionary creation, accessing values, common methods, iteration, and dictionary comprehension.

**How to use:** Run each code cell in order. Read the explanation before running the examples.

## 1. What is a dictionary?
A dictionary stores information as **key-value pairs**. Think of a key as a label and its value as the information attached to that label.

Example: `"name"` is a key and `"Alice"` is its value.

Dictionaries are **mutable** (you can change them), keys must be **unique**, and modern Python dictionaries preserve insertion order (Python 3.7+).

## 2. Creating dictionaries

```python
# An empty dictionary
empty_dict = {}
print(empty_dict)

# A dictionary containing key-value pairs
student = {
    "name": "Alice",
    "age": 25,
    "grade": "A"
}
print(student)

# Creating a dictionary with dict()
person = dict(name="John", age=30, city="New York")
print(person)
```

## 3. Accessing dictionary elements
- `dictionary[key]` returns the value, but raises a `KeyError` if the key does not exist.
- `get(key)` returns `None` if the key is missing.
- `get(key, default)` returns your chosen default value if the key is missing.

```python
print(student["name"])          # Alice
print(student.get("age"))       # 25
print(student.get("height"))    # None
print(student.get("height", "Not Found"))  # Not Found
```

## 4. Adding and updating values

```python
# Add a new key-value pair
student["city"] = "New York"
print(student)

# Update an existing value
student["age"] = 26
print(student)
```

## 5. Common dictionary methods

| Method | What it does |
|---|---|
| `keys()` | Provides the dictionary's keys |
| `values()` | Provides the dictionary's values |
| `items()` | Provides key-value pairs as tuples |
| `get(key, default)` | Gets a value safely, with an optional default |
| `update(other)` | Adds or updates pairs from another dictionary |
| `pop(key, default)` | Removes a key and returns its value |
| `popitem()` | Removes and returns the last inserted pair |
| `setdefault(key, default)` | Gets a value; if missing, inserts the default |
| `clear()` | Removes all items |
| `copy()` | Makes a shallow copy |

**Note:** The examples below use separate dictionaries where helpful, so each method's effect is easy to see.

### `keys()`, `values()`, and `items()`

```python
student = {"name": "Alice", "age": 26, "grade": "A"}
print(student.keys())
print(student.values())
print(student.items())
```

### `get()`

```python
print(student.get("age", 0))       # 26
print(student.get("height", 0))    # 0 because height is missing
```

### `update()`

```python
student.update({"age": 27, "city": "New York"})
print(student)
```

### `pop()`

```python
removed_grade = student.pop("grade")
print("Removed value:", removed_grade)
print(student)

# A missing key can use a default instead of raising KeyError
print(student.pop("height", "Not Found"))
```

### `popitem()`

```python
example = {"name": "Alice", "age": 25, "grade": "A"}
removed_pair = example.popitem()
print("Removed pair:", removed_pair)
print(example)
```

### `setdefault()`

```python
example = {"name": "Alice", "age": 25}
print(example.setdefault("age", 0))   # Existing value: 25
print(example.setdefault("city", "Unknown"))  # Inserts and returns Unknown
print(example)
```

### `copy()` and `clear()`

```python
original = {"name": "Alice", "age": 25}
new_dict = original.copy()
new_dict["age"] = 99
print("Original:", original)
print("Copy:", new_dict)

temporary = original.copy()
temporary.clear()
print("After clear():", temporary)
```

## 6. Removing an item with `del`
Python also supports `del dictionary[key]`. It removes the pair but does not return its value.

```python
example = {"name": "Alice", "age": 25, "grade": "A"}
del example["grade"]
print(example)
```

## 7. Iterating over a dictionary

```python
student = {"name": "Alice", "age": 26, "city": "New York"}

# Loop through keys
for key in student:
    print(key)

# Loop through values
for value in student.values():
    print(value)

# Loop through both keys and values
for key, value in student.items():
    print(key, ":", value)
```

## 8. Dictionary comprehension
A dictionary comprehension creates a dictionary in a compact way.

Pattern: `{key_expression: value_expression for item in iterable}`

```python
squares = {x: x**2 for x in range(1, 6)}
print(squares)
# {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}
```

## 9. Important properties
- **Ordered:** Python 3.7+ guarantees insertion order.
- **Mutable:** You can add, update, and remove pairs.
- **Unique keys:** Repeating a key overwrites its earlier value.
- **Hashable keys:** Common key types include strings, numbers, and tuples containing only hashable values. Lists and dictionaries cannot be keys.
- **Flexible values:** Values can be strings, numbers, lists, other dictionaries, or other Python objects.

```python
# Duplicate keys: the later value replaces the earlier one
example = {"age": 20, "age": 21}
print(example)  # {'age': 21}

# Values can be different types
mixed = {"name": "Alice", "scores": [85, 92], "active": True}
print(mixed)
```

## Quick recap
- Use `[]` to access a required key and `get()` when a key may be missing.
- Use assignment to add or change a value.
- Use `update()` to add or change multiple pairs.
- Use `pop()` to remove a known key and `popitem()` to remove the last inserted pair.
- Use `items()` to loop over keys and values together.
- Use dictionary comprehension to build dictionaries from an iterable.
