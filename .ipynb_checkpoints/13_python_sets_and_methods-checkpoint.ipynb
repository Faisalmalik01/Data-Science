# Sets and Set Methods

A set in Python is an **unordered, mutable, and unique** collection of elements. It does not allow duplicate values.

## Creating a Set

```python
# Empty set (must use set(), not {} — {} creates an empty dict!)
empty_set = set()

# Set with elements
numbers = {1, 2, 3, 4, 5}

# Mixed data types
mixed_set = {1, "Hello", 3.14, True}

# Creating a set from a list — duplicates are removed automatically
unique_numbers = set([1, 2, 2, 3, 4, 4, 5])
print(unique_numbers)  # {1, 2, 3, 4, 5}
```

## Key Properties of Sets

- **Unordered**: No indexing or slicing.
- **Unique Elements**: Duplicates are automatically removed.
- **Mutable**: You can add or remove elements.

```python
my_set = {10, 20, 30}

# my_set[0]          # ❌ TypeError: 'set' object is not subscriptable
# print(my_set[0:2]) # ❌ No slicing either

print(20 in my_set)  # ✅ Membership testing works (and is very fast)
```

## Common Set Methods

| Method | Description | Example |
|---|---|---|
| `add(x)` | Adds an element `x` to the set | `my_set.add(10)` |
| `update(iterable)` | Adds multiple elements from an iterable | `my_set.update([6, 7, 8])` |
| `remove(x)` | Removes `x` (**raises an error** if not found) | `my_set.remove(3)` |
| `discard(x)` | Removes `x` (**no error** if not found) | `my_set.discard(3)` |
| `pop()` | Removes and returns a **random** element | `my_set.pop()` |
| `clear()` | Removes all elements | `my_set.clear()` |
| `copy()` | Returns a shallow copy | `my_set.copy()` |

```python
my_set = {1, 2, 3}

my_set.add(4)                  # Add one element
print(my_set)

my_set.update([5, 6, 7])       # Add multiple elements
print(my_set)

my_set.discard(99)             # No error, even though 99 isn't there
print(my_set)

my_set.remove(1)               # Remove (would raise KeyError if missing)
print(my_set)

copy = my_set.copy()
my_set.clear()
print("Cleared:", my_set, "| Copy survived:", copy)
```

## Set Operations

| Operation | Operator | Description |
|---|---|---|
| `union(set2)` | `\|` | All unique elements from both sets |
| `intersection(set2)` | `&` | Elements common to both sets |
| `difference(set2)` | `-` | Elements in `set1` but **not** in `set2` |
| `symmetric_difference(set2)` | `^` | Elements in either set, but **not both** |
| `issubset(set2)` | `<=` | `True` if `set1` is a subset of `set2` |
| `issuperset(set2)` | `>=` | `True` if `set1` is a superset of `set2` |

```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

# Union – combines all unique elements
print(set1 | set2)              # {1, 2, 3, 4, 5, 6}
print(set1.union(set2))         # same result

# Intersection – common elements
print(set1 & set2)              # {3, 4}
print(set1.intersection(set2))  # same result

# Difference – in set1 but not in set2
print(set1 - set2)              # {1, 2}
print(set1.difference(set2))    # same result

# Symmetric Difference – in either set, but not both
print(set1 ^ set2)              # {1, 2, 5, 6}
print(set1.symmetric_difference(set2))  # same result
```

```python
# Subset / Superset checks
a = {1, 2, 3}
b = {1, 2, 3, 4, 5}

print(a.issubset(b))    # True — every element of a is in b
print(b.issuperset(a))  # True — b contains all of a
print(a <= b, b >= a)   # Same checks with operators
```

## In-Place Set Operations

Adding `=` updates the set directly instead of returning a new one (`|=`, `&=`, `-=`, `^=`).

```python
s = {1, 2}
s |= {3, 4}     # Same as s.update({3, 4})
print(s)        # {1, 2, 3, 4}

s &= {2, 3, 5}  # Same as s.intersection_update(...)
print(s)        # {2, 3}
```

## Set Comprehension

```python
squares = {x ** 2 for x in range(6)}
print(squares)  # {0, 1, 4, 9, 16, 25}
```

## Frozen Sets (`frozenset`)

An **immutable** version of a set. Just like tuples vs. lists, a frozen set can be used as a **dictionary key** or stored inside another set.

```python
frozen = frozenset([1, 2, 3])
# frozen.add(4)   # ❌ AttributeError: 'frozenset' object has no attribute 'add'

# Usable as dictionary keys and inside other sets
valid_pairs = {frozenset(["a", "b"]): "pair AB", frozenset(["c", "d"]): "pair CD"}
print(valid_pairs[frozenset(["a", "b"])])

set_of_sets = {frozenset([1, 2]), frozenset([3, 4])}
print(set_of_sets)
```

## Sets in Data Science

### Example: Deduplication and Comparing Datasets

```python
# Customers from two months
jan_customers = {"Alice", "Bob", "Charlie", "Diana"}
feb_customers = {"Bob", "Diana", "Eve", "Frank"}

print("Loyal (both months):", jan_customers & feb_customers)
print("Churned (Jan only):", jan_customers - feb_customers)
print("New (Feb only):", feb_customers - jan_customers)
print("All customers:", jan_customers | feb_customers)

# Quick deduplication of raw data
raw_ids = [101, 102, 101, 103, 102, 104]
unique_ids = list(set(raw_ids))
print("Unique IDs:", unique_ids)
```

## Summary

- Sets are **unordered, mutable, and unique** — no duplicates, no indexing.
- Use `discard()` when the element might not exist; `remove()` otherwise.
- `|` (union), `&` (intersection), `-` (difference), `^` (symmetric difference).
- `frozenset` is the immutable variant — usable as dict keys.
- In data science: fast deduplication, membership tests, and dataset comparison.
