# File Handling in Python

Beginner-friendly notes and runnable examples for creating, reading, writing, appending, and deleting files.

**Run cells from top to bottom.** These examples create files in the notebook's current working directory.

## 1. What is file handling?

Variables usually store data while a program runs. Files let us store data on disk and use it later. File handling means opening, reading, writing, updating, or deleting files.

Python uses the built-in `open()` function to open files.

## 2. Opening a file

Syntax: `open(filename, mode)`

- **Filename:** for example, `example.txt`
- **Mode:** specifies what you want to do with the file.

```python
# Create a sample file so the reading examples work
with open("example.txt", "w", encoding="utf-8") as file:
    file.write("Hello from Python!\nThis is file handling.")
print("File created.")
```

## 3. File modes

| Mode | Meaning |
|---|---|
| `r` | Read; file must exist |
| `w` | Write; creates a file or **overwrites** existing content |
| `a` | Append to the end; creates the file if needed |
| `x` | Create a new file; errors if it already exists |
| `b` | Binary mode, used with `rb`, `wb`, etc. |
| `t` | Text mode; default, used with `rt`, `wt`, etc. |

**Be careful:** opening a file in `w` mode erases its previous contents.

## 4. Reading with `read()`

`read()` reads the remaining file content into one string.

```python
with open("example.txt", "r", encoding="utf-8") as file:
    content = file.read()
print(content)
```

## 5. Reading one line with `readline()`

Each call to `readline()` reads the next line. `repr()` makes newline characters visible.

```python
with open("example.txt", "r", encoding="utf-8") as file:
    line1 = file.readline()
    line2 = file.readline()
print("First line:", repr(line1))
print("Second line:", repr(line2))
```

## 6. Reading all lines with `readlines()`

`readlines()` returns a list of strings, usually keeping each line's `\n` character.

```python
with open("example.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()
print(lines)
print("Number of lines:", len(lines))
```

## 7. Writing with `write()`

Mode `w` overwrites existing content. `write()` does not add a newline automatically.

```python
with open("write_demo.txt", "w", encoding="utf-8") as file:
    file.write("Hello, World!\n")
    file.write("This is a second line.\n")

with open("write_demo.txt", "r", encoding="utf-8") as file:
    print(file.read())
```

## 8. Writing multiple lines with `writelines()`

`writelines()` writes each string, but it does **not** add newline characters for you.

```python
lines_to_write = ["Hello\n", "Welcome to Python\n", "File Handling\n"]
with open("writelines_demo.txt", "w", encoding="utf-8") as file:
    file.writelines(lines_to_write)

with open("writelines_demo.txt", "r", encoding="utf-8") as file:
    print(file.read())
```

## 9. Appending with mode `a`

Append mode keeps the old content and adds new content at the end.

```python
with open("example.txt", "a", encoding="utf-8") as file:
    file.write("\nThis is an additional line.")

with open("example.txt", "r", encoding="utf-8") as file:
    print(file.read())
```

## 10. Why use `with open()`?

`with open(...) as file:` automatically closes the file when the indented block ends, even if an error occurs. This is safer than remembering to call `file.close()` yourself.

```python
with open("example.txt", "r", encoding="utf-8") as file:
    content = file.read()

# The file is closed here, but the string remains available.
print(content)
```

## 11. Checking whether a file exists

```python
import os

filename = "example.txt"
if os.path.exists(filename):
    print("File exists!")
else:
    print("File not found!")
```

## 12. Deleting a file

`os.remove()` deletes a file. Check the path carefully before deleting anything.

```python
import os

filename = "delete_me.txt"
# Create a demo file first
with open(filename, "w", encoding="utf-8") as file:
    file.write("Temporary deletion demo")

if os.path.exists(filename):
    os.remove(filename)
    print("File deleted.")
else:
    print("File does not exist.")
```

## 13. Binary files

Images, PDFs, and other non-text files should be handled in binary mode such as `rb` (read binary) and `wb` (write binary). This example uses a small byte sequence, so no image file is required.

```python
sample_bytes = bytes([65, 66, 67, 0, 255])

with open("sample.bin", "wb") as file:
    file.write(sample_bytes)

with open("sample.bin", "rb") as file:
    data = file.read()
print(data)

with open("sample_copy.bin", "wb") as file:
    file.write(data)
print("Binary file copied.")
```

## 14. Quick reference

| Operation | Example |
|---|---|
| Open for reading | `open("example.txt", "r")` |
| Read all content | `file.read()` |
| Read one line | `file.readline()` |
| Read lines as a list | `file.readlines()` |
| Write / overwrite | `open("example.txt", "w")` then `file.write("Hello")` |
| Write several strings | `file.writelines(lines)` |
| Append | `open("example.txt", "a")` |
| Check existence | `os.path.exists("file.txt")` |
| Delete | `os.remove("file.txt")` |
| Read binary | `open("sample.bin", "rb")` |

For ordinary text files, prefer `with open(filename, mode, encoding="utf-8") as file:`.

## 15. Practice questions

Try these before running the solutions.

1. Create `notes.txt` and write `I am learning Python.` to it.
2. Read and print its contents.
3. Append `File handling is useful.` on a new line.
4. Print each line using a `for` loop.
5. Check whether `notes.txt` exists.
6. Write three students' names and scores from a dictionary into a text file.

### Practice solutions

```python
# 1. Create and write
with open("notes.txt", "w", encoding="utf-8") as file:
    file.write("I am learning Python.\n")

# 2. Read
with open("notes.txt", "r", encoding="utf-8") as file:
    print(file.read())

# 3. Append
with open("notes.txt", "a", encoding="utf-8") as file:
    file.write("File handling is useful.\n")

# 4. Print each line
with open("notes.txt", "r", encoding="utf-8") as file:
    for line in file:
        print(line, end="")

# 5. Check existence
import os
print(os.path.exists("notes.txt"))

# 6. Write dictionary entries
scores = {"Aisha": 90, "Bilal": 84, "Sara": 95}
with open("scores.txt", "w", encoding="utf-8") as file:
    for name, score in scores.items():
        file.write(f"{name}: {score}\n")

with open("scores.txt", "r", encoding="utf-8") as file:
    print(file.read())
```

## Final recap

- `r` reads, `w` overwrites, `a` appends, and `x` creates only if the file does not exist.
- `read()`, `readline()`, and `readlines()` read content in different ways.
- `write()` and `writelines()` do not add newline characters automatically.
- Use `with open()` to close files automatically.
- Use `rb` and `wb` for binary data.
- Check file paths carefully before deleting files.
